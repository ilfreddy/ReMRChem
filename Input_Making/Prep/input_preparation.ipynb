```python
mask = open("input_mask.txt")
content = mask.read()
mask.close()
#
box_size = 1  # This depends on the nuclear charge!
potential_type = "fermi_dirac"
d1_d2 = [False, True]
prec_list = [4,5,6,7,8]
atom_label = "Hg"
atom_charge = "80"
molecule_list = [[atom_label, atom_charge, "0.0", "0.0", "0.0", "54", "54"] ,
#                 ["He", "2", "0.1", "0.2", "0.3", "1.0e-4", "3.5722217300e8"] ,
#                 ["He", "2", "0.0", "0.0", "0.0", "1.0e-5", "3.5722217300e10"],
                 [atom_label, atom_charge, "0.1", "0.2", "0.3", "54", "54"]]

deriv = 'BS'
content = content.replace('DERIVATIVE', deriv)
content = content.replace('TWOELECTRON','False')
content = content.replace('POT_TYPE',potential_type)
content = content.replace('BOX_SIZE',str(box_size))

for p in prec_list:
    p_set = content.replace('P_PREC', str(p))
    p_set = p_set.replace('P_THR', str(p-1))
    p_set = p_set.replace('ORDER', str(p+3))
    for m in molecule_list:
        mol_set = p_set.replace('ATOM',     m[0])
        mol_set = mol_set.replace('CHARGE',   m[1])
        mol_set = mol_set.replace('XPOS',     m[2])
        mol_set = mol_set.replace('YPOS',     m[3])
        mol_set = mol_set.replace('ZPOS',     m[4])
        mol_set = mol_set.replace('RADIUS',   m[5]) #Not used in the current implementation
        mol_set = mol_set.replace('EXPONENT', m[6])
        for d in d1_d2:
            if(d): 
                d_set = mol_set.replace('D1_OR_D2', 'True')
                print("D is true ", d)
                d_set = d_set.replace('READORB', 'True')
                d_set = d_set.replace('SAVEORB', 'False')
                d_set = d_set.replace('READPOT', 'True')
                d_set = d_set.replace('SAVEPOT', 'False')
                d_set = d_set.replace('COMPPOT', 'False')
            else:
                d_set = mol_set.replace('D1_OR_D2', 'False')
                print("D is false", d)
                d_set = d_set.replace('READORB', 'False')
                d_set = d_set.replace('SAVEORB', 'True')
                d_set = d_set.replace('READPOT', 'False')
                d_set = d_set.replace('SAVEPOT', 'True')
                d_set = d_set.replace('COMPPOT', 'True')
            if m[2] == "0.0" and m[3] == "0.0" and m[4] == "0.0":
                centered = 1
            else:
                centered = 0

            index = str(centered) + str(int(d)) + str(p) + deriv
            name = "input__"+atom_label+str(index)+".txt"
            print(name)
            file = open(name,"w")
            file.write(d_set)
            file.close()
            
```

```python
def extract_numbers(file_path, search_string, nl):
    numbers = []
    
    with open(file_path, 'r') as file:
        lines = file.readlines()
        total_lines = len(lines)

        for i in range(total_lines):
            if search_string in lines[i]:
                for j in range(0, nl):
                    if i + j < total_lines:
                        next_line_number = float(lines[i + j].split()[-1])
                        numbers.append(next_line_number)

    return numbers  # Return whatever we have, which may be less than 3

# Example usage

search_string = 'Final E'  # Replace with your unique strings

#search_path = "/Users/lfr001/src/ReMRChem/argon_gaussian_pot_derPH/"
search_path = ""


#nr_of_lines = 3
#for i in range(16):
#    filename = search_path + 'out' + str(i) + '.txt'
#    numbers = extract_numbers(filename, search_string, nr_of_lines)
#    print(numbers[0], numbers[1], numbers[2])

nr_of_lines = 5
search_string = 'Dirac E'  # Replace with your unique strings
for i in range(0,16):
    filename = search_path + 'out' + str(i) + '.txt'
    numbers = extract_numbers(filename, search_string, nr_of_lines)
    print(numbers[0], numbers[1], numbers[2], numbers[3], numbers[4])

```
